```python
from langchain_community.tools.tavily_search import TavilySearchResults 
search = TavilySearchResults()
```

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_community.vectorstores import FAISS 
from langchain_text_splitters import RecursiveCharacterTextSplitter

loader = WebBaseLoader(["https://www.wikipedia.org/wiki/Langchain"])
docs = loader.load()
```

```python
docs
```

```python

documents = RecursiveCharacterTextSplitter(
    chunk_size=1000, chunk_overlap=200
).split_documents(docs)


vector = FAISS.from_documents(documents,embedding)
retriever = vector.as_retriever()
```

```python
retriever.invoke("What is Langchain?")
```

```python
from langchain.tools.retriever import create_retriever_tool

retriever_tool = create_retriever_tool(retriever,"langchain_studio","Search information about langchain ,for any questions about langchain ,you must use this tool to get the information")
```

```python
tool = [search,retriever_tool]

```

```python
from langchain.agents import create_tool_calling_agent

agent = create_tool_calling_agent(llm,tool,prompt)
```

```python
from langchain.agents import AgentExecutor

agent_executor = AgentExecutor(agent=agent, tools=tool, llm=llm)
```

```python
agent_executor.invoke({"input": "What is Langchain and who created it?"})
```
